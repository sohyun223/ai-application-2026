## 2. Python 빠른 점검

### 2-1. 변수 (Variable)
변수는 값을 담는 **이름표**입니다. `=` 는 "같다"가 아니라 "담는다"라는 뜻입니다.

````
# 변수: 값에 이름을 붙인다
distance_km = 12.5     # 통학 거리 (실수, float)
speed_kmh = 25         # 버스 평균 속도 (정수, int)
route_name = "7016번 버스"  # 문자열 (str)

# 변수로 계산하기
time_min = distance_km / speed_kmh * 60

# f-string: 문자열 안에 {변수}를 끼워 넣는 문법 (매우 자주 씁니다)
print(f"노선: {route_name}")
print(f"거리 {distance_km} km, 평균 {speed_kmh} km/h")
print(f"예상 소요 시간: {time_min:.0f}분")  # :.0f = 소수점 없이 표시
````

### 2-2. 리스트 (List)

리스트는 여러 값을 **순서대로** 담는 자료구조입니다. 데이터 분석의 출발점!

- 대괄호 `[ ]`로 만들고, **인덱스(index)는 0부터** 셉니다.
- `len()` 으로 개수, `.append()` 로 추가.

````
# 자율주행차의 센서 목록
sensors = ["카메라", "레이더", "LiDAR", "GPS"]

print(sensors[0])        # 첫 번째 (0번) 항목 -> 카메라
print(sensors[-1])       # 마지막 항목 -> GPS
print(len(sensors), "개") # 개수

sensors.append("초음파")   # 항목 추가
print(sensors)

# 숫자 리스트: 어느 도로의 구간별 평균 속도 (km/h)
speeds = [42, 55, 38, 60, 47]
print(f"최고 속도: {max(speeds)} km/h, 최저 속도: {min(speeds)} km/h")
print(f"평균 속도: {sum(speeds) / len(speeds):.1f} km/h")
````

### 2-3. for 반복문

리스트의 항목을 **하나씩 꺼내며** 같은 작업을 반복합니다.

```
for 변수 in 리스트:
    반복할 코드   # <- 들여쓰기(스페이스 4칸)가 문법입니다!
```

````
# 구간별 속도를 하나씩 점검하며 정체 여부 표시
speeds = [42, 55, 38, 60, 47]  # 구간 1~5의 평균 속도

for i, s in enumerate(speeds):          # enumerate: 번호(i)와 값(s)을 함께
    if s < 40:
        status = "정체"
    elif s < 50:
        status = "서행"
    else:
        status = "원활"
    print(f"구간 {i + 1}: {s:>3} km/h  ->  {status}")

# range(n): 0부터 n-1까지의 숫자를 만들어 주는 함수
total = 0
for h in range(3):
    total += 100  # total = total + 100
print(f"3시간 동안 통과 차량: {total}대 (가정)")
````

### 2-4. 함수 (Function)

함수는 **자주 쓰는 코드에 이름을 붙여 재사용**하는 도구입니다.

- `def 이름(입력):` 으로 정의, `return` 으로 결과 반환
- 앞으로 배울 **AI 모델도 결국 '입력 → 출력' 함수**라는 점, 슬라이드에서 강조했지요!

````
def classify_congestion(speed_kmh):
    """평균 속도(km/h)를 받아 교통 상태 문자열을 돌려주는 함수"""
    if speed_kmh < 40:
        return "정체"
    elif speed_kmh < 50:
        return "서행"
    else:
        return "원활"

# 함수 호출: 이름(입력값)
print(classify_congestion(35))   # -> 정체
print(classify_congestion(45))   # -> 서행
print(classify_congestion(70))   # -> 원활

# 리스트 + for + 함수를 한꺼번에!
speeds = [42, 55, 38, 60, 47]
labels = [classify_congestion(s) for s in speeds]   # 리스트 컴프리헨션 (2주차에 자세히)
print(labels)
````

## 3. 데이터 시각화 맛보기 — 시간대별 교통량

**가상의(synthetic)** 시간대별 교통량 데이터를 만들어 그래프로 그려 봅니다.

- `numpy` : 숫자 배열 계산 라이브러리 (2주차에 배움)
- `matplotlib` : 그래프 라이브러리 — 시간에 따른 **추이**는 **선 그래프(line chart)**가 정석
- 도시 교통량의 특징: **출근(8시)·퇴근(18시) 이중 피크(twin peaks)**

> 그래프의 글자는 Colab 한글 폰트 문제를 피해 영문으로 표기합니다. 코드의 의미는 2주차부터 차근차근 배우니, 오늘은 **실행해서 그림이 나오는 경험**이 목표입니다.

````
import numpy as np

rng = np.random.default_rng(42)   # 난수 고정: 누가 실행해도 같은 결과

hours = np.arange(24)             # 0, 1, 2, ..., 23시

# 출근(8시)·퇴근(18시) 피크를 가우시안 봉우리로 흉내 낸 합성 데이터
base = (800
        + 900 * np.exp(-((hours - 8) ** 2) / 4)     # 출근 피크
        + 1100 * np.exp(-((hours - 18) ** 2) / 5)   # 퇴근 피크
        - 500 * np.exp(-((hours - 3) ** 2) / 6))    # 새벽 감소
traffic = (base + rng.normal(0, 60, 24)).round().astype(int)

# 표(table)로 먼저 눈으로 확인 - 그래프 전에 데이터부터 보는 습관!
print("시간대 | 교통량 (대/시)")
print("-" * 24)
for h in [0, 4, 8, 12, 16, 18, 22]:
    print(f"{h:>4}시  | {traffic[h]:>6,}")
print(f"\n하루 최대: {traffic.max():,}대 ({traffic.argmax()}시)")
````

````
import matplotlib.pyplot as plt

# 단일 계열 선 그래프: 시간의 흐름 -> 선(line)이 정석
plt.figure(figsize=(9, 4))
plt.plot(hours, traffic, color="#2563A8", linewidth=2,
         marker="o", markersize=6)

# 피크 지점에만 선택적으로 값 표시 (모든 점에 숫자를 붙이면 지저분!)
peak = traffic.argmax()
plt.annotate(f"{traffic[peak]:,}", (peak, traffic[peak]),
             textcoords="offset points", xytext=(0, 10),
             ha="center", fontsize=9, color="#333333")

plt.title("Hourly Traffic Volume - Synthetic Data (Week 01 Demo)")
plt.xlabel("Hour of Day")
plt.ylabel("Vehicles / hour")
plt.xticks(range(0, 24, 2))
plt.grid(alpha=0.3)          # 눈금선은 은은하게
plt.tight_layout()
plt.show()

print("출근(8시)과 퇴근(18시) 두 개의 봉우리가 보이면 성공!")
````

### 3-1. 두 계열 비교: 평일 vs 주말

주말은 출퇴근 피크가 사라지고 **낮 시간대가 완만하게** 붐빕니다.

두 개 이상의 선을 그릴 때의 규칙:
- 계열마다 **다른 색** + **다른 마커(marker)** (색약자도 구분 가능하도록)
- **범례(legend)** 필수 — 어느 선이 무엇인지

# [3-1] 두 계열 비교: 평일 vs 주말
#  - 주말은 출퇴근 피크가 사라지고 낮 시간대가 완만하게 붐빕니다.
#  - 두 계열 이상 그릴 때: 계열마다 다른 색 + 다른 마커(색약자 배려), 범례(legend) 필수!

# 주말 데이터: 피크 없이 낮에 완만하게 증가하는 형태로 합성
weekend = (700
           + 650 * np.exp(-((hours - 14) ** 2) / 30)
           - 450 * np.exp(-((hours - 4) ** 2) / 8))
weekend = (weekend + rng.normal(0, 50, 24)).round().astype(int)

plt.figure(figsize=(9, 4))
plt.plot(hours, traffic, color="#2563A8", linewidth=2,
         marker="o", markersize=6, label="Weekday")
plt.plot(hours, weekend, color="#B05C10", linewidth=2,
         marker="s", markersize=6, label="Weekend")

plt.title("Weekday vs Weekend Traffic - Synthetic Data")
plt.xlabel("Hour of Day")
plt.ylabel("Vehicles / hour")
plt.xticks(range(0, 24, 2))
plt.grid(alpha=0.3)
plt.legend()                 # 범례: 두 계열 이상이면 필수
plt.tight_layout()
plt.show()

print("관찰: 평일은 8시/18시 이중 피크, 주말은 14시 부근 단일 완만 피크")
